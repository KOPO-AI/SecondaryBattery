### 변수와 자료형 — 데이터에 이름과 종류가 있다 `필수`

```python
voltage = 3.72          # 셀 전압 (V)
capacity = 2850         # 셀 용량 (mAh)
lot_id = "LOT-2408-A17" # 로트 번호
is_pass = True          # 합격 여부
print(voltage, capacity, lot_id, is_pass)
print(type(voltage), type(capacity), type(lot_id), type(is_pass))

# 교재 실행 결과 ------------------------------------------------------
#   3.72 2850 LOT-2408-A17 True
#   <class 'float'> <class 'int'> <class 'str'> <class 'bool'>
```

### 실수 오차와 문자열 다루기 `필수`

```python
print(3.65 - 3.7)              # 부동소수점 오차
lot_id = "LOT-2408-A17"
print(lot_id.replace("-", "_"))
print(lot_id.split("-"))
print(len(lot_id))

# 교재 실행 결과 ------------------------------------------------------
#   -0.050000000000000266
#   LOT_2408_A17
#   ['LOT', '2408', 'A17']
#   12
```

### f-string: 보고서 문장 만들기 `필수`

```python
voltage = 3.7215
lot_id = "LOT-2408-A17"
print(f"로트 {lot_id}의 평균 전압은 {voltage:.2f} V입니다.")
print(f"용량: {2850:,} mAh")
print(f"|{voltage:>8.2f}|{'합격':>6}|")   # 폭 8칸 오른쪽 정렬

# 교재 실행 결과 ------------------------------------------------------
#   로트 LOT-2408-A17의 평균 전압은 3.72 V입니다.
#   용량: 2,850 mAh
#   |    3.72|    합격|
```

### 형변환 `필수`

```python
raw = "3.72"            # 장비가 넘겨준 문자열
voltage = float(raw)    # 실수로 변환
print(voltage + 0.05)
print(int(3.99))        # 소수점 이하 버림
print(str(2850) + " mAh")

# 교재 실행 결과 ------------------------------------------------------
#   3.77
#   3
#   2850 mAh
```
