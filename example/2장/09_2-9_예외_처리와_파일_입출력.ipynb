### 대표 사례 2 — 형변환 실패 (ValueError) `필수`

```python
values = ["3607.4", "3603.1", "오류", "3598.3"]
clean = []
for v in values:
    try:
        clean.append(float(v))
    except ValueError:
        print(f"숫자로 바꿀 수 없는 값 발견: {v!r} → 건너뜀")
print(clean)

# 교재 실행 결과 ------------------------------------------------------
#   숫자로 바꿀 수 없는 값 발견: '오류' → 건너뜀
#   [3607.4, 3603.1, 3598.3]
```

### with open — 파일을 안전하게 열고 닫기 `필수`

```python
with open("data/battery_process_data.csv", "r", encoding="utf-8-sig") as f:
    for i, line in enumerate(f):
        print(line.strip())
        if i >= 2:
            break

# 교재 실행 결과 ------------------------------------------------------
#   Lot_ID,믹싱_RPM,믹싱_온도,코팅_토출압력,건조로_1구간_온도,프레스_압력,프레스_Gap,최종_용량_mAh,전극_면저항_mOhmcm2,불량_여부
#   L25000,1774,24.9,116.4,115.0,41.0,93.5,3607.4,1428.4,0
#   L25001,1667,23.9,114.7,112.8,43.6,93.2,3603.1,1444.9,0
```
