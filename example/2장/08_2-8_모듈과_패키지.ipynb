### 자주 쓰는 표준 라이브러리 네 가지 `선택 학습(자율 복습)`

### 나만의 유틸 모듈 만들기 — process_utils.py `선택 학습(자율 복습)`

```python
%%writefile process_utils.py
"""공정 데이터 처리용 공용 함수 모음 (process_utils.py)"""


def calc_defect_rate(defect_count, total_count):
    """불량률(%)을 계산한다."""
    if total_count == 0:
        return 0.0
    return defect_count / total_count * 100
```

```python
#  교재 2.8.5절: "새 셀에서 import process_utils 를 실행한 뒤
#  print(process_utils.calc_defect_rate(3, 250)) 을 부르면 1.2 가 출력된다."
#  example 폴더에서 python "2장/08_2-8_모듈과_패키지.py" 로 실행하면 스크립트가 있는
#  2장/ 폴더가 import 경로에 자동으로 들어가므로 같은 폴더의 process_utils.py 를 찾는다.
import process_utils
print(process_utils.calc_defect_rate(3, 250))

# 교재 실행 결과 ------------------------------------------------------
#   1.2
```
